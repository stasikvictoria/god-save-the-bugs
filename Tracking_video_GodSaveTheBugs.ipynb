```python
from ultralytics import YOLO

model = YOLO("last.pt")
results = model.track(source="IMG_5357.mov", show=True, tracker="bytetrack.yaml", save=True, conf=0.5)
```

```python
result = results[0]
box = result.boxes
face_count = len(box)
```

```python
import pandas as pd
import numpy as np
import datetime

df = pd.DataFrame(columns = ['Species', 'Confiance', 'id_video', 'Time'])

for j in results:
  result = j
  box = result.boxes
  for box in result.boxes:
      class_id = result.names[box.cls[0].item()]
      conf = round(box.conf[0].item(), 2)
      conf = float(conf)
      id = box.id
      if id == None:
        id = 0
      else:
        id = id
      if id == 0:
        id = 0
      else:
        id = int(id.item())
      print("Type:", class_id, conf, id)
      time = datetime.datetime.today()
      date_formatee = time.strftime("%d/%m/%Y %H:%M")
      df.loc[j] = [class_id, conf, id, date_formatee]
```

```python
df = df.reset_index()
```

```python
df = df.drop(['index'], axis = 1)
```

```python
df
```

```python
df2 = df.drop(df[df['id_video'] == 0].index, inplace = False)
```

```python
df2
```

```python
df3 = df2.groupby(['id_video','Species','Time'], as_index=False).mean('Confiance')
df3
```

```python
data = [['ladybug', 'coleopteres'], ['bee', 'pollinisateur'], ['butterfly', 'pollinisateur']]

df_data = pd.DataFrame(data, columns =['Species', 'Category'])
```

```python
df_test = df3.merge(df_data, how='inner', on='Species')
df_test
```

```python
df_final = df_test[['Species','Category','Time']]
df_final
```
