# Imports

```python
import cv2
from ultralytics import YOLO
import os
from tqdm.notebook import tqdm
```

# Initializing and training YOLO model

```python
# Load a model
yolo = YOLO("yolov8n.yaml")  # build a new model from scratch

# Use the model
model = yolo.train(data="config.yaml", epochs=50)  # train the model
```

# Setting up video input and output

```python
video_path_out = '{}_out.mp4'.format('video.mp4') # sets the output video path based on the input video file name
frames = cv2.VideoCapture('video.mp4') # creates a VideoCapture object to read frames from the video file "video.mp4"
video_not_finished, frame = frames.read() # video_not_finished is a boolean value indicating whether the frame was successfully read or not
H, W, _ = frame.shape
out = cv2.VideoWriter(video_path_out, cv2.VideoWriter_fourcc(*'MP4V'), int(frames.get(cv2.CAP_PROP_FPS)), (W, H)) # creates a VideoWriter object to write the processed frames to the output video file
```

# Loading the trained model

```python
model_path = os.path.join('.', 'runs', 'detect', 'train', 'weights', 'last.pt')
model = YOLO(model_path)  # load a custom model
threshold = 0.5
class_name_dict = {0: 'bee', 1: 'ladybug'}
```

# Object detection and visualization

```python
while video_not_finished:

    results = model(frame)[0] # YOLO model is applied to the frame

    #  iterates over each detected object and draws bounding boxes and labels on the frame using OpenCV
    for result in results.boxes.data.tolist():
        x1, y1, x2, y2, score, class_id = result

        if score > threshold:
            cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 4)
            cv2.putText(frame, class_name_dict[int(class_id)].upper(), (int(x1), int(y1 - 10)),
                        cv2.FONT_HERSHEY_SIMPLEX, 1.3, (0, 255, 0), 3, cv2.LINE_AA)

    # processed frame is written to the output video
    out.write(frame)
    video_not_finished, frame = frames.read()
```

# Cleaning up

```python
frames.release()
out.release()
cv2.destroyAllWindows()
```
